## Ejercicio 3: Inspección visual de un producto
### Ficha PEAS
- **Percepción (S):** matriz de 5x5 (NumPy) donde cada valor es 0 (pixel normal) o 1 (pixel defectuoso)
- **Acciones (A):** aprobar el producto, enviar a revisión manual, rechazar el producto
- **Entorno (E):** línea de producción, totalmente observable dentro del recorte de imagen capturado (el agente ve completamente la matriz que recibe)
- **Objetivo:** detectar productos defectuosos antes de que salgan de la línea de producción, minimizando tanto los defectos no detectados como los descartes innecesarios de productos buenos
- **Medida de desempeño (P):** minimizar productos defectuosos que pasan como aprobados y minimizar productos buenos rechazados o enviados a revisión sin necesidad


### Justificación de las reglas
El número de píxeles defectuosos es un indicador directo y objetivo de la calidad de la imagen
capturada, por lo que se usa como criterio principal. Cero defectos indica un producto en buen
estado y se aprueba automáticamente. Entre 1 y 3 defectos, el margen de duda es razonable: puede
tratarse de ruido en la imagen o un defecto real menor, por lo que se prefiere una revisión
manual en lugar de decidir automáticamente. Más de 3 defectos ya representa una cantidad
suficientemente alta como para rechazar el producto sin necesidad de intervención humana,
priorizando así no dejar pasar productos claramente defectuosos.
