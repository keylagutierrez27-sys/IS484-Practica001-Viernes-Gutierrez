## Ejercicio 2: Escalamiento de tickets de soporte técnico
### Ficha PEAS
- **Percepción (S):** tiempo_espera_minutos, nivel_urgencia, cliente_premium
- **Acciones (A):** mantener en nivel 1, escalar a nivel 2, escalar a nivel 3
- **Entorno (E):** cola de tickets de mesa de ayuda, parcialmente observable (no se conoce la complejidad real del problema hasta que un agente humano lo revisa)
- **Objetivo:** resolver los tickets en el menor tiempo posible priorizando los casos más urgentes y a los clientes con mayor nivel de servicio contratado
- **Medida de desempeño (P):** minimizar el tiempo de espera de tickets urgentes y maximizar la satisfacción de clientes premium, sin saturar innecesariamente los niveles superiores de soporte


### Justificación de las reglas

La urgencia alta se escala de inmediato porque, por definición, representa un riesgo o impacto
que no puede esperar, independientemente de otros factores. La urgencia media se trata de forma
condicional: si el tiempo de espera ya es alto o el cliente es premium, se escala para evitar que
el problema empeore o que un cliente de alto valor se sienta desatendido; si no se cumple ninguna
de esas condiciones, se mantiene en nivel 1 para no saturar innecesariamente los niveles
superiores. La urgencia baja normalmente se queda en nivel 1, salvo un caso especial: un cliente
premium con una espera excesivamente larga, donde el compromiso comercial con ese cliente
justifica una excepción a la regla general.


```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    if nivel_urgencia == "alta":
        return "escalar a nivel 3", "urgencia alta reportada por el cliente"

    elif nivel_urgencia == "media":
        if tiempo_espera_minutos > 30 or cliente_premium:
            return "escalar a nivel 2", "urgencia media con espera prolongada o cliente premium"
        else:
            return "mantener en nivel 1", "urgencia media sin condiciones agravantes"

    else:  # nivel_urgencia == "baja"
        if cliente_premium and tiempo_espera_minutos > 60:
            return "escalar a nivel 2", "cliente premium con espera excesiva pese a urgencia baja"
        else:
            return "mantener en nivel 1", "urgencia baja sin agravantes"
```

### Simulación y pruebas

```python
casos = [
    (10, "alta", False),     # se escala de inmediato por urgencia alta
    (45, "media", False),    # espera > 30 -> escalar a nivel 2
    (15, "media", True),     # premium -> escalar a nivel 2 aunque la espera sea corta
    (10, "media", False),    # sin agravantes -> se mantiene en nivel 1
    (70, "baja", True),      # premium + espera excesiva -> escala aunque la urgencia sea baja
    (65, "baja", False),     # urgencia baja, no premium, espera larga -> se mantiene en nivel 1
]

for tiempo, urgencia, premium in casos:
    accion, motivo = agente_soporte(tiempo, urgencia, premium)
    print(f"Espera: {tiempo}min | Urgencia: {urgencia} | Premium: {premium} -> {accion} ({motivo})")
```
